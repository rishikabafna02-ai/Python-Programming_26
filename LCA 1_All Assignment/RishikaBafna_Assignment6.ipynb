```python
import os
import pandas as pd

# ---------- Step 0: Create a sample file (skip if you have yo) ----------
sample = """ID,Name,Age,Department,Marks,City,Remarks
1,Asha,20,CSE,85,Pune,Good
2,Rohan,21,IT,78,Mumbai,Good
3,Meera,,CSE,92,Pune,Excellent
4,Karan,22,ENTC,,Nashik,Average
5,Sneha,20,IT,88,Pune,Good
2,Rohan,21,IT,78,Mumbai,Good
6,Vikram,23,CSE,67,,Average
7,Anita,,ENTC,74,Mumbai,Good
8,Zoya,21,IT,,Pune,Average
9,Dev,22,CSE,95,Nashik,Excellent
"""
file_name = "students.csv"          # change to your own .csv / .xls / .xlsx file
if not os.path.exists(file_name):
    with open(file_name, "w") as f:
        f.write(sample)

# ---------- Step 1: Load the data (CSV or Excel) ----------
if file_name.endswith(".csv"):
    df = pd.read_csv(file_name)
else:
    df = pd.read_excel(file_name)    # needs: pip install openpyxl (xlsx) / xlrd (xls)

print("=== 1. Original Data ===")
print(df)
print("\nShape (rows, columns):", df.shape)
print("\nData types and non-null counts:")
df.info()

# ---------- Step 2: Detect missing values ----------
print("\n=== 2. Missing Values per Column ===")
print(df.isnull().sum())

# ---------- Step 3: Duplicate records ----------
print("\n=== 3. Duplicate Records ===")
print("Number of duplicate rows:", df.duplicated().sum())
print(df[df.duplicated()])

df = df.drop_duplicates()            # keep the first occurrence
print("Shape after removing duplicates:", df.shape)

# ---------- Step 4: Remove irrelevant columns ----------
df = df.drop(columns=["Remarks"])    # not needed for analysis
print("\nColumns after drop():", list(df.columns))

# ---------- Step 5: Handle missing values ----------
# Option A: dropna() - removes rows/columns containing missing values
rows_dropped = df.dropna()
print("\n=== 5A. dropna() ===")
print("Rows left if every row with a missing value is dropped:", len(rows_dropped))

# Option B: fillna() - replaces missing values (keeps all the data)
df["Age"] = df["Age"].fillna(df["Age"].median())        # numeric -> median
df["Marks"] = df["Marks"].fillna(df["Marks"].median())  # numeric -> median
df["City"] = df["City"].fillna(df["City"].mode()[0])    # text -> most frequent

print("\n=== 5B. After fillna() ===")
print(df.isnull().sum())

# Fix data types after filling
df["Age"] = df["Age"].astype(int)
df["Marks"] = df["Marks"].astype(int)

# ---------- Step 6: Selection and filtering using loc[] ----------
print("\n=== 6. Selection and Filtering with loc[] ===")

# (a) Select specific columns for all rows
print("\n(a) Name and Marks of all students:")
print(df.loc[:, ["Name", "Marks"]])

# (b) Filter rows by a condition
print("\n(b) Students with Marks > 80:")
print(df.loc[df["Marks"] > 80, ["Name", "Department", "Marks"]])

# (c) Multiple conditions
print("\n(c) CSE students aged 21 or above:")
print(df.loc[(df["Department"] == "CSE") & (df["Age"] >= 21)])

# (d) Select by row label range
print("\n(d) Rows with index 0 to 2:")
print(df.loc[0:2])

# ---------- Step 7: Save the cleaned data ----------
df.to_csv("students_cleaned.csv", index=False)
print("\nCleaned data saved to students_cleaned.csv")
```
