```python
#Different Operations on List,Tupple and Dictionary data structures.
# Problem Statement- Write a python program to create a Dictionary,Tupple, List of Students and perform the following operations:Add,Delete,Update

#LIST
                          # ADD OPERATOR

Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
Students.append("Tanishka")
print(Students)
Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
Students.insert( 0,"Tanishka")
print(Students)
```

```python
                      #DELETE OPERATOR

Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
Students.remove("Vansh")
print(Students)
Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
Students.pop(5)
print(Students)
Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
del Students[4]
print(Students)
```

```python
                    #UPDATE OPERATOR

Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
Students[1:3] = ["Tanishka", "Akhilesh"]
print(Students)
Students= ["Aditya","Rishika","Om","Vansh","Viraaj","Rachita"]
Students[3] = "Janhavi"
print(Students)
```

```python
#TUPPLE
                  #ADD OPERATOR
Students= ("Aditya","Rishika","Om","Vansh","Viraaj","Rachita")
print(Students)
Students= ("Aditya","Rishika","Om","Vansh","Viraaj","Rachita")
y = list(Students)
y.append("Tanishka")
Students = tuple(y)
print(Students)
```

```python
                #DELETE OPERATOR

Students= ("Aditya","Rishika","Om","Vansh","Viraaj","Rachita")
y = list(Students)
y.remove("Rachita")
Students = tuple(y)
print(Students)
```

```python
                #UPDATE OPERATOR

Students= ("Aditya","Rishika","Om","Vansh","Viraaj","Rachita")
y = list(Students)
y[4] = "Soham"
Students = tuple(y)
print(Students)
```

```python
#DICTIONARY
               #ADD OPERATOR

Students= {
    "Aditya" : 123456,
    "Rishika": 654321,
    "Om"     : 414141,
}
Students["Vansh"]= 242424
print(Students)
```

```python
             #UPDATE OPERATOR

Students= {
    "Aditya" : 123456,
    "Rishika": 654321,
    "Om"     : 414141,
}
Students.update({"Vansh":242424})
print(Students)
```

```python
            #DELETE OPERATOR
Students= {
    "Aditya" : 123456,
    "Rishika": 654321,
    "Om"     : 414141,
}
Students.pop("Rishika")
print(Students)
```
